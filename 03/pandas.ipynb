## 미니퀘스트 병합 1번

## 미니퀘스트 병합 3번

```python
import pandas as pd

df1 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '대전'],
    '구매': [10000, 20000, 30000]
})

df2 = pd.DataFrame({
    '고객ID': [1, 2, 3],
    '도시': ['서울', '부산', '광주'],
    '구매': [15000, 25000, 35000]
})

result = pd.merge(
    df1,
    df2,
    on=['고객ID', '도시'],
    how='inner',
    suffixes=('_df1', '_df2')
)

print(result)
```

## 피벗 미니퀘스트 2번

```python
import pandas as pd

data = {
    '카테고리': ['전자', '가전', '전자', '가전'],
    '제품': ['A', 'B', 'A', 'B'],
    '판매량': [100, 200, 150, 250]
}

df = pd.DataFrame(data)

result = pd.pivot_table(
    df,
    index='카테고리',
    columns='제품',
    values='판매량',
    aggfunc='sum'
)

print(result)
```
